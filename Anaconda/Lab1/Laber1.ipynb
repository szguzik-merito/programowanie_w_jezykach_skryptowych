## Instrukcje skoku

```python
for i in range(10):  # Przerywa pętlę (break) kiedy i=5
    if i == 5:
        break
    print(i)
```

```python
for i in range(10):  # Przechodzi do print kiedy reszte z dzielenia i przez 2 wynosi 0
    if i % 2 == 0:
        continue
    print(i)
```
